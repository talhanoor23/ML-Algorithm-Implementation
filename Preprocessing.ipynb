```python
from sklearn.model_selection import train_test_split
```

```python
train_val_df, test_df= train_test_split(weather_df, test_size=0.2, random_state=42)
train_df, val_df= train_test_split(train_val_df, test_size=0.25, random_state=42)
```

```python
print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

```python
plt.title('year wise count')
plt.xlabel('year')
plt.ylabel('count')
sns.countplot(x=pd.to_datetime(weather_df.Date).dt.year);
```

here we can divide data according to date, not randomly bcz date play imp role


```python
year = pd.to_datetime(weather_df.Date).dt.year
```

```python
train_df = weather_df[year < 2015]
val_df = weather_df[year == 2015]
test_df = weather_df[year > 2015]
```

```python
print(train_df.shape)
print(val_df.shape)
print(test_df.shape)
```

```python
input_col = list(train_df.columns)[1:-1]
output_col = 'RainTomorrow'
```

```python
train_inputs = train_df[input_col]
train_outputs = train_df[output_col]
val_inputs = val_df[input_col]
val_outputs = val_df[output_col]
test_inputs = test_df[input_col]
test_outputs = test_df[output_col]
```

```python
numeric_train_inputs = train_inputs.select_dtypes(include=np.number).columns.tolist()
#categorical_train_inputs = train_inputs.select_dtypes(exclude=np.number).columns.tolist()
categorical_train_inputs = train_inputs.select_dtypes('object').columns.tolist()
```

```python
train_inputs[numeric_train_inputs].describe()
```

```python
train_inputs[categorical_train_inputs].nunique()
```

```python
weather_df[numeric_train_inputs].isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
```

```python
imputer.fit(weather_df[numeric_train_inputs])
```

```python
imputer.statistics_
```

```python
train_inputs[numeric_train_inputs] = imputer.transform(train_inputs[numeric_train_inputs])
val_inputs[numeric_train_inputs] = imputer.transform(val_inputs[numeric_train_inputs])
test_inputs[numeric_train_inputs] = imputer.transform(test_inputs[numeric_train_inputs])
```

```python
train_inputs[numeric_train_inputs].isnull().sum()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
scaler.fit(weather_df[numeric_train_inputs])
```

```python
train_inputs[numeric_train_inputs] = scaler.transform(train_inputs[numeric_train_inputs])
val_inputs[numeric_train_inputs] = scaler.transform(val_inputs[numeric_train_inputs])
test_inputs[numeric_train_inputs] = scaler.transform(test_inputs[numeric_train_inputs])
```

```python
train_inputs[numeric_train_inputs]
```

```python
train_inputs[numeric_train_inputs].describe()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
weather_df2 = weather_df[categorical_train_inputs].fillna('Unknown')
```

```python
encoder.fit(weather_df2[categorical_train_inputs])
```

```python
encoder.categories_
```

```python
encoded_cols = list(encoder.get_feature_names_out(categorical_train_inputs))
print(encoded_cols)
```

```python
train_inputs[encoded_cols] = encoder.transform(train_inputs[categorical_train_inputs].fillna('Unknown'))
val_inputs[encoded_cols] = encoder.transform(val_inputs[categorical_train_inputs].fillna('Unknown'))
test_inputs[encoded_cols] = encoder.transform(test_inputs[categorical_train_inputs].fillna('Unknown'))
```
